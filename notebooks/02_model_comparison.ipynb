# Model Comparison

Comparing Prophet (baseline/regressors) with XGBoost.

```
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt

# Load MLflow runs from local SQLite database
conn = sqlite3.connect('../mlruns/mlflow.db')
query = """
SELECT 
    runs.name as run_name,
    m_mae.value as MAE,
    m_rmse.value as RMSE,
    m_mape.value as MAPE,
    m_r2.value as R2
FROM runs
LEFT JOIN metrics m_mae ON runs.run_uuid = m_mae.run_uuid AND m_mae.key = 'MAE'
LEFT JOIN metrics m_rmse ON runs.run_uuid = m_rmse.run_uuid AND m_rmse.key = 'RMSE'
LEFT JOIN metrics m_mape ON runs.run_uuid = m_mape.run_uuid AND m_mape.key = 'MAPE'
LEFT JOIN metrics m_r2 ON runs.run_uuid = m_r2.run_uuid AND m_r2.key = 'R2'
WHERE runs.status = 'FINISHED'
"""
df_metrics = pd.read_sql(query, conn)
conn.close()

# Display Model Comparison Table
df_metrics.sort_values(by='MAE')
```
