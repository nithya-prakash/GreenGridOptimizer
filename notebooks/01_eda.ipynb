# Exploratory Data Analysis

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from pathlib import Path

# Load Data
data_path = Path('../data/processed/master_dataset.parquet')
df = pd.read_parquet(data_path)
df.head()
```

```
# Generation Time Series
fig, ax = plt.subplots(figsize=(15, 6))
df[['wind_onshore', 'wind_offshore', 'solar']].plot(ax=ax)
plt.title("Renewable Generation Time Series")
plt.ylabel("MW")
plt.show()
```

```
# Seasonality Patterns (Daily)
df['hour'] = df.index.hour
hourly_avg = df.groupby('hour')[['solar', 'wind_onshore', 'wind_offshore']].mean()
hourly_avg.plot(figsize=(10, 5), title="Average Daily Profile")
plt.ylabel("MW")
plt.show()
```

```
# Missing Value Summary
print("Missing Value Counts:")
print(df.isna().sum())
```

```
# Weather Distributions
df[['wind_speed_10m', 'temperature_2m', 'cloud_cover', 'shortwave_radiation']].hist(figsize=(12, 8), bins=30)
plt.suptitle("Weather Distributions")
plt.show()
```

```
# Correlation Heatmap
plt.figure(figsize=(10, 8))
corr = df[['wind_onshore', 'solar', 'wind_speed_10m', 'shortwave_radiation', 'cloud_cover', 'temperature_2m']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation between Weather and Generation")
plt.show()
```
